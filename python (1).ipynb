```python
#loops = A loops is used to repeat a block of code again until a certain condition is met.
#For Loop :- It is used when you know the no. of iterations you'll use a FOR loop.
list1 = [1,2,4,6,4]
for i in list1:
  print(i)
```

```python
#WHILE loop :- is used when you don't know how many iteration you've to use but you know a condition that determine when to stop you'll use WHILE loop.
i = 0
while i<5:
  print(i, end = " ")
  i += 1
```
